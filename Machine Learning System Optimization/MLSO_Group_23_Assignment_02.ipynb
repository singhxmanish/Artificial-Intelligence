```python
# ============================================================
# 0. FORCE SAME PYTHON FOR DRIVER & WORKERS
# ============================================================
import os, sys, time, numpy as np

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

print("Python used:", sys.executable)
print("Python version:", sys.version)


# ============================================================
# 1. IMPORTS
# ============================================================
from pyspark.sql import SparkSession
from pyspark.ml.feature import VectorAssembler
from pyspark.sql.functions import col
from pyspark.ml.evaluation import BinaryClassificationEvaluator

from xgboost.spark import SparkXGBClassifier
import kagglehub
import matplotlib.pyplot as plt


# ============================================================
# 2. CREATE SPARK SESSION (LOCAL MULTI-CORE)
# ============================================================
spark = (
    SparkSession.builder
    .master("local[8]")                 # 8 CPU cores on Mac
    .appName("XGBoost-Spark-Local")
    .config("spark.sql.execution.arrow.pyspark.enabled", "true")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("WARN")

print("Spark master:", spark.sparkContext.master)
print("Default parallelism:", spark.sparkContext.defaultParallelism)


# ============================================================
# 3. LOAD DATA
# ============================================================
path = kagglehub.dataset_download("mlg-ulb/creditcardfraud")
csv_path = os.path.join(path, "creditcard.csv")

df = spark.read.csv(csv_path, header=True, inferSchema=True)

feature_cols = [c for c in df.columns if c != "Class"]
assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")

df = assembler.transform(df).select("features", "Class")

train_df, test_df = df.randomSplit([0.8, 0.2], seed=42)


# ============================================================
# 4. METRICS FUNCTION
# ============================================================
def classification_metrics(preds):
    TP = preds.filter((col("Class")==1) & (col("prediction")==1)).count()
    TN = preds.filter((col("Class")==0) & (col("prediction")==0)).count()
    FP = preds.filter((col("Class")==0) & (col("prediction")==1)).count()
    FN = preds.filter((col("Class")==1) & (col("prediction")==0)).count()

    acc = (TP+TN)/(TP+TN+FP+FN)
    prec = TP/(TP+FP) if TP+FP else 0
    rec = TP/(TP+FN) if TP+FN else 0
    f1 = (2*prec*rec)/(prec+rec) if prec+rec else 0
    return acc, prec, rec, f1


# ============================================================
# 5. SERIAL TRAINING (1 PARTITION)
# ============================================================
def train_serial(df):
    df = df.repartition(1).cache()

    model = SparkXGBClassifier(
        features_col="features",
        label_col="Class",
        tree_method="hist",
        max_depth=6,
        n_estimators=100,
        learning_rate=0.1,
        eval_metric="auc"
    )

    start = time.time()
    trained_model = model.fit(df)
    end = time.time()

    return trained_model, end - start


# ============================================================
# 6. PARALLEL TRAINING (SPARK DATA PARALLELISM)
# ============================================================
def train_parallel(df):
    df = df.repartition(8).cache()   # Spark-level parallelism

    model = SparkXGBClassifier(
        features_col="features",
        label_col="Class",

        # IMPORTANT:
        # Spark controls parallelism → DO NOT set nthread
        num_workers=1,               # local machine
        tree_method="hist",
        max_depth=6,
        n_estimators=100,
        learning_rate=0.1,
        eval_metric="auc"
    )

    start = time.time()
    trained_model = model.fit(df)
    end = time.time()

    return trained_model, end - start


# ============================================================
# 7. RUN EXPERIMENTS
# ============================================================
spark.catalog.clearCache()

serial_model, T_serial = train_serial(train_df)
parallel_model, T_parallel = train_parallel(train_df)

speedup = T_serial / T_parallel


# ============================================================
# 8. EVALUATION
# ============================================================
preds = parallel_model.transform(test_df)

acc, prec, rec, f1 = classification_metrics(preds)

evaluator = BinaryClassificationEvaluator(
    labelCol="Class",
    rawPredictionCol="probability",
    metricName="areaUnderROC"
)
auc = evaluator.evaluate(preds)


# ============================================================
# 9. RESULTS
# ============================================================
print("\n===== FINAL RESULTS =====")
print(f"T_serial   : {T_serial:.2f} s")
print(f"T_parallel : {T_parallel:.2f} s")
print(f"Speedup    : {speedup:.2f} x")
print(f"Accuracy   : {acc:.4f}")
print(f"Precision  : {prec:.4f}")
print(f"Recall     : {rec:.4f}")
print(f"F1-score   : {f1:.4f}")
print(f"AUC-ROC    : {auc:.4f}")


# ============================================================
# 10. SPEEDUP PLOT
# ============================================================
plt.bar(["Serial", "Spark Parallel"], [T_serial, T_parallel])
plt.ylabel("Training Time (s)")
plt.title("Spark Data Parallel XGBoost Speedup")
plt.show()
```
