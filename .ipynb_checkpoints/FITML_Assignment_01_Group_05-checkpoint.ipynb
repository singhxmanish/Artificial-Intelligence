<p style="font-size: 16px;"><b><b>Task 2:</b> Using 5 fold cross-validation, implement a Naïve Bayes classifier with all attributes and report the group-level fairness metrics (demographic parity, equality of odds) and accuracy for protected variable, gender. No preprocessing should be applied. [1.5 marks]</p>

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, confusion_matrix

# ---------------------------------------------------------
# 1. Load and Prepare Data
# ---------------------------------------------------------
df = pd.read_csv('clean_dataset.csv')

# Encode categorical columns to numbers so Naive Bayes can run.
le = LabelEncoder()

categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    df[col] = le.fit_transform(df[col])

# Define Features (X) and Target (y)
X = df.drop('Approved', axis=1)
y = df['Approved']
sensitive_attr = df['Gender']  # Protected/Sensitive attribute

# ---------------------------------------------------------
# 2. Initialize 5-Fold Cross-Validation & Classifier
# ---------------------------------------------------------
kf = KFold(n_splits=5, shuffle=True, random_state=42)
nb_classifier = GaussianNB()

# Lists to store metrics for each fold
accuracies = []
dp_diffs = []        # Demographic Parity Differences
tpr_diffs = []       # True Positive Rate Differences (Eq. of Odds TPR part )
fpr_diffs = []       # False Positive Rate Differences (Eq. of Odd FPR part )

print("Running Task 2: Running 5-Fold Cross-Validation...\n")

# ---------------------------------------------------------
# 3. Loop through Folds
# ---------------------------------------------------------
fold = 1
for train_idx, test_idx in kf.split(X):
    # Split data
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    gender_test = sensitive_attr.iloc[test_idx] # Gender labels for test set
    
    # Train Model
    nb_classifier.fit(X_train, y_train)
    
    # Predict
    y_pred = nb_classifier.predict(X_test)
    
    # --- A. Accuracy ---
    acc = accuracy_score(y_test, y_pred)
    accuracies.append(acc)
    
    # Create a temporary dataframe for fairness calculations
    res = pd.DataFrame({'y_true': y_test, 'y_pred': y_pred, 'Gender': gender_test})
    
    # --- B. Demographic Parity (Statistical Parity) ---
    # Formula: P(pred=1 | G=1) - P(pred=1 | G=0)
    prob_1 = res[res['Gender'] == 1]['y_pred'].mean()
    prob_0 = res[res['Gender'] == 0]['y_pred'].mean()
    dp_diffs.append(prob_1 - prob_0)
    
    # --- C. Equality of Odds ---
    # Check True Positive Rate (TPR) and False Positive Rate (FPR) parity
    def get_rates(df_group):
        cm = confusion_matrix(df_group['y_true'], df_group['y_pred'], labels=[0, 1])
        tn, fp, fn, tp = cm.ravel()
        tpr = tp / (tp + fn) if (tp + fn) > 0 else 0
        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0
        return tpr, fpr

    tpr_1, fpr_1 = get_rates(res[res['Gender'] == 1])
    tpr_0, fpr_0 = get_rates(res[res['Gender'] == 0])
    
    tpr_diffs.append(tpr_1 - tpr_0)
    fpr_diffs.append(fpr_1 - fpr_0)
    
    fold += 1

# ---------------------------------------------------------
# 4. Report Average Results
# ---------------------------------------------------------
print("-" * 40)
print(f"Average Accuracy: {np.mean(accuracies):.4f}")
print("-" * 40)
print("Group-Level Fairness Metrics (Gender 1 - Gender 0):")
print(f"Demographic Parity Difference:      {np.mean(dp_diffs):.4f}")
print(f"Equality of Odds (TPR Difference):  {np.mean(tpr_diffs):.4f}")
print(f"Equality of Odds (FPR Difference):  {np.mean(fpr_diffs):.4f}")
print("-" * 40)
```

<p style="font-size: 16px; color: blue;"><b>Authors Note Task 2:</b> Average Accuracy - 78.84% <br> <br> <b> Demographic Parity </b> This value is very close to 0. It indicates that the probability of getting a positive prediction (loan approval) is almost the same for Gender 1 and Gender 0. Specifically, Gender 1 has a slightly lower probability of being predicted "Approved". <br> <br> <b> Equality of Odds </b> Both differences are extremely small (less than 1%). This suggests the model satisfies Equality of Odds quite well; it is correctly identifying qualified candidates (TPR) and incorrectly identifying unqualified candidates (FPR) at almost identical rates for both genders <br> <br> The baseline Naïve Bayes model appears to be quite fair with respect to gender on this dataset, showing very little bias in both demographic parity and equality of odds metrics.</p>

<p style="font-size: 16px;"><b><b>Task 4:</b> Incorporate reweighting preprocessing and report the accuracy and group-level fairness metrics using 5 fold cross validation for the Naïve Bayes Classifier. [1.5 marks]</p>



```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, confusion_matrix

# ---------------------------------------------------------
# 1. Load and Prepare Data
# ---------------------------------------------------------
df = pd.read_csv('clean_dataset.csv')

# Encode categorical columns
le = LabelEncoder()
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    df[col] = le.fit_transform(df[col])

X = df.drop('Approved', axis=1)
y = df['Approved']

# ---------------------------------------------------------
# 2. Define Reweighting Function
# ---------------------------------------------------------
def compute_sample_weights(df_train, sensitive_col, target_col):
    """
    Computes weights for each sample to make sensitive attribute 
    and target independent.
    Weight = (P(S) * P(Y)) / P(S, Y)
    """
    n_total = len(df_train)
    n_s = df_train[sensitive_col].value_counts()
    n_y = df_train[target_col].value_counts()
    n_sy = df_train.groupby([sensitive_col, target_col]).size()
    
    weights = []
    for i, row in df_train.iterrows():
        s = row[sensitive_col]
        t = row[target_col]
        
        # Calculate weight
        # W = (Count(S) * Count(Y)) / (Total * Count(S,Y))
        numer = n_s[s] * n_y[t]
        denom = n_total * n_sy.get((s, t), 0)
        
        w = numer / denom if denom != 0 else 1.0
        weights.append(w)
        
    return np.array(weights)

# ---------------------------------------------------------
# 3. Initialize 5-Fold Cross-Validation
# ---------------------------------------------------------
kf = KFold(n_splits=5, shuffle=True, random_state=42)
nb_classifier = GaussianNB()

accuracies = []
dp_diffs = []
tpr_diffs = []
fpr_diffs = []

print("Running Task 4 (Reweighting Preprocessing)...\n")

# ---------------------------------------------------------
# 4. Train and Evaluate
# ---------------------------------------------------------
for train_idx, test_idx in kf.split(X):
    # Split data
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    
    # --- Apply Reweighting on Training Data ---
    # We combine X and y temporarily to compute weights based on the distribution
    train_data = X_train.copy()
    train_data['Approved'] = y_train
    
    # Compute weights
    sample_weights = compute_sample_weights(train_data, 'Gender', 'Approved')
    
    # Train Model with Weights
    nb_classifier.fit(X_train, y_train, sample_weight=sample_weights)
    
    # Predict (Test data is NOT reweighted, as we want real-world performance)
    y_pred = nb_classifier.predict(X_test)
    gender_test = X_test['Gender']
    
    # --- Metrics ---
    accuracies.append(accuracy_score(y_test, y_pred))
    
    # Fairness DataFrame
    res = pd.DataFrame({'y_true': y_test, 'y_pred': y_pred, 'Gender': gender_test})
    
    # Demographic Parity
    prob_1 = res[res['Gender'] == 1]['y_pred'].mean()
    prob_0 = res[res['Gender'] == 0]['y_pred'].mean()
    dp_diffs.append(prob_1 - prob_0)
    
    # Equality of Odds
    def get_rates(df_group):
        cm = confusion_matrix(df_group['y_true'], df_group['y_pred'], labels=[0, 1])
        tn, fp, fn, tp = cm.ravel()
        tpr = tp / (tp + fn) if (tp + fn) > 0 else 0
        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0
        return tpr, fpr

    tpr_1, fpr_1 = get_rates(res[res['Gender'] == 1])
    tpr_0, fpr_0 = get_rates(res[res['Gender'] == 0])
    
    tpr_diffs.append(tpr_1 - tpr_0)
    fpr_diffs.append(fpr_1 - fpr_0)

# ---------------------------------------------------------
# 5. Report Results
# ---------------------------------------------------------
print("-" * 40)
print(f"Average Accuracy: {np.mean(accuracies):.4f}")
print("-" * 40)
print("Group-Level Fairness Metrics (Gender 1 - Gender 0):")
print(f"Demographic Parity Difference:      {np.mean(dp_diffs):.4f}")
print(f"Equality of Odds (TPR Difference):  {np.mean(tpr_diffs):.4f}")
print(f"Equality of Odds (FPR Difference):  {np.mean(fpr_diffs):.4f}")
print("-" * 40)
```

<p style="font-size: 16px; color: blue;"><b>Authors Note Task 4:</b> The accuracy is ideally same as the baseline model in Task 2 (~78.84%). This tells applying reweighting  to mitigate bias did not harm the overall predictive performance of the model. <br> <br> <b> Demographic Parity</b> The gap has narrowed slightly (closer to 0) <br> <br> <b>Equality of Odds</b>: <br>
TPR Difference: ~-0.0020 (previously -0.0061)<br>
FPR Difference: ~0.0020 (previously 0.0047). <br> The True Positive Rate difference became smaller (improved), meaning the model is even more consistent in approving qualified candidates across genders. The False Positive Rate difference increased slightly but remains very low (less than 1%)</p>

<p style="font-size: 16px;"><b><b>Task 5:</b> Incorporate universal sampling preprocessing and report the accuracy and group-level fairness metrics using 5 fold cross validation for the Naïve Bayes Classifier. [1.5 marks]</p>



```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, confusion_matrix

# ---------------------------------------------------------
# 1. Load and Prepare Data
# ---------------------------------------------------------
df = pd.read_csv('clean_dataset.csv')

# Encode categorical columns
le = LabelEncoder()
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    df[col] = le.fit_transform(df[col])

X = df.drop('Approved', axis=1)
y = df['Approved']

# ---------------------------------------------------------
# 2. Define Sampling Logic
# ---------------------------------------------------------
def apply_uniform_sampling(X_train, y_train, sensitive_col='Gender'):
    """
    Resamples the dataset so that all 4 subgroups (Gender x Approved)
    have the same number of samples (equal to the size of the largest subgroup).
    """
    train_df = X_train.copy()
    train_df['Approved'] = y_train
    
    # Identify the 4 subgroups
    g0_y0 = train_df[(train_df[sensitive_col] == 0) & (train_df['Approved'] == 0)]
    g0_y1 = train_df[(train_df[sensitive_col] == 0) & (train_df['Approved'] == 1)]
    g1_y0 = train_df[(train_df[sensitive_col] == 1) & (train_df['Approved'] == 0)]
    g1_y1 = train_df[(train_df[sensitive_col] == 1) & (train_df['Approved'] == 1)]
    
    groups = [g0_y0, g0_y1, g1_y0, g1_y1]
    
    # Determine target size (max group size) to avoid data loss
    max_size = max(len(g) for g in groups)
    
    # Resample each group to the target size
    resampled_groups = [g.sample(max_size, replace=True, random_state=42) for g in groups]
    
    # Combine and shuffle
    train_df_balanced = pd.concat(resampled_groups)
    train_df_balanced = train_df_balanced.sample(frac=1, random_state=42).reset_index(drop=True)
    
    return train_df_balanced.drop('Approved', axis=1), train_df_balanced['Approved']

# ---------------------------------------------------------
# 3. Initialize Cross-Validation
# ---------------------------------------------------------
kf = KFold(n_splits=5, shuffle=True, random_state=42)
nb_classifier = GaussianNB()

accuracies = []
dp_diffs = []
tpr_diffs = []
fpr_diffs = []

print("Running Task 5 (Universal/Uniform Sampling)...\n")

# ---------------------------------------------------------
# 4. Train and Evaluate
# ---------------------------------------------------------
for train_idx, test_idx in kf.split(X):
    # Split data
    X_train_orig, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train_orig, y_test = y.iloc[train_idx], y.iloc[test_idx]
    
    # --- Apply Universal Sampling on Training Data ---
    X_train_bal, y_train_bal = apply_uniform_sampling(X_train_orig, y_train_orig)
    
    # Train Model on Balanced Data
    nb_classifier.fit(X_train_bal, y_train_bal)
    
    # Predict on Original Test Data (Evaluation must be on real distribution)
    y_pred = nb_classifier.predict(X_test)
    gender_test = X_test['Gender']
    
    # --- Metrics ---
    accuracies.append(accuracy_score(y_test, y_pred))
    
    # Fairness DataFrame
    res = pd.DataFrame({'y_true': y_test, 'y_pred': y_pred, 'Gender': gender_test})
    
    # Demographic Parity
    prob_1 = res[res['Gender'] == 1]['y_pred'].mean()
    prob_0 = res[res['Gender'] == 0]['y_pred'].mean()
    dp_diffs.append(prob_1 - prob_0)
    
    # Equality of Odds
    def get_rates(df_group):
        cm = confusion_matrix(df_group['y_true'], df_group['y_pred'], labels=[0, 1])
        tn, fp, fn, tp = cm.ravel()
        tpr = tp / (tp + fn) if (tp + fn) > 0 else 0
        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0
        return tpr, fpr

    tpr_1, fpr_1 = get_rates(res[res['Gender'] == 1])
    tpr_0, fpr_0 = get_rates(res[res['Gender'] == 0])
    
    tpr_diffs.append(tpr_1 - tpr_0)
    fpr_diffs.append(fpr_1 - fpr_0)

# ---------------------------------------------------------
# 5. Report Results
# ---------------------------------------------------------
print("-" * 40)
print(f"Average Accuracy: {np.mean(accuracies):.4f}")
print("-" * 40)
print("Group-Level Fairness Metrics (Gender 1 - Gender 0):")
print(f"Demographic Parity Difference:      {np.mean(dp_diffs):.4f}")
print(f"Equality of Odds (TPR Difference):  {np.mean(tpr_diffs):.4f}")
print(f"Equality of Odds (FPR Difference):  {np.mean(fpr_diffs):.4f}")
print("-" * 40)
```

<p style="font-size: 16px; color: blue;"><b>Authors Note Task 5:</b> Interestingly, the accuracy increased significantly compared to the baseline and previous tasks (which were around ~78.8%). This means balancing the classes (oversampling the minority cases) prevented the model from being biased towards the majority class (rejected applicants) <br> <br> <b> Demographic Parity: </b> -0.0401 The difference has increased (in magnitude), meaning the gap between the two groups has widened slightly in favor of Gender 0 <br> <br> <b>Equality of Odds</b>: <br>
TPR Difference: ~-0.0491 (previously -0.0061)<br>
FPR Difference: ~-0.0049 (previously 0.0047). <br> The TPR difference also increased slightly to ~-5%. This means Gender 0 applicants who should be approved are being approved at a slightly higher rate than Gender 1 applicants who should be approved. <br> <br> Universal sampling provided the best accuracy among all methods tested so far.</p></p>

<p style="font-size: 16px;"><b><b>Task 6:</b> Incorporate preferential sampling based preprocessing and report the accuracy and group-level fairness metrics using 5 fold cross validation for the Naïve Bayes Classifier. [2 marks]</p>

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, confusion_matrix

# ---------------------------------------------------------
# 1. Load and Prepare Data
# ---------------------------------------------------------
df = pd.read_csv('clean_dataset.csv')

# Encode categorical columns
le = LabelEncoder()
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    df[col] = le.fit_transform(df[col])

X = df.drop('Approved', axis=1)
y = df['Approved']

# ---------------------------------------------------------
# 2. Define Preferential Sampling Logic
# ---------------------------------------------------------
def apply_preferential_sampling(X_train, y_train, sensitive_col='Gender', target_col='Approved'):
    """
    Implements Preferential Sampling by duplicating 'borderline' samples 
    to shift the decision boundary.
    """
    # Step A: Train a ranker to get probabilities
    ranker = GaussianNB()
    ranker.fit(X_train, y_train)
    probs = ranker.predict_proba(X_train)[:, 1]
    
    # Create temp dataframe
    df_temp = X_train.copy()
    df_temp[target_col] = y_train
    df_temp['prob'] = probs
    
    # Step B: Identify Candidates for Resampling
    
    # 1. Unprivileged Positives (Gender 0, Approved 1)
    # Sort by probability ascending -> so that those  with LOW prob (closest to boundary)
    deprived_positives = df_temp[(df_temp[sensitive_col] == 0) & (df_temp[target_col] == 1)]
    deprived_positives = deprived_positives.sort_values(by='prob', ascending=True)
    
    # 2. Privileged Negatives (Gender 1, Approved 0)
    # Sort by probability descending -> sso that those with HIGH prob (closest to boundary)
    privileged_negatives = df_temp[(df_temp[sensitive_col] == 1) & (df_temp[target_col] == 0)]
    privileged_negatives = privileged_negatives.sort_values(by='prob', ascending=False)

    # Step C: Duplicate borderline cases top 50 

    n_dupe_dp = int(len(deprived_positives) * 0.5)
    n_dupe_pn = int(len(privileged_negatives) * 0.5)
    
    dupe_dp = deprived_positives.head(n_dupe_dp)
    dupe_pn = privileged_negatives.head(n_dupe_pn)
    
    # Step D: Construct new training set
    df_resampled = pd.concat([df_temp, dupe_dp, dupe_pn])
    df_resampled = df_resampled.drop('prob', axis=1) # Clean up
    
    # Shuffle
    df_resampled = df_resampled.sample(frac=1, random_state=42).reset_index(drop=True)
    
    return df_resampled.drop(target_col, axis=1), df_resampled[target_col]

# ---------------------------------------------------------
# 3. Cross-Validation Execution
# ---------------------------------------------------------
kf = KFold(n_splits=5, shuffle=True, random_state=42)
nb_classifier = GaussianNB()

accuracies = []
dp_diffs = []
tpr_diffs = []
fpr_diffs = []

print("Running Task 6 (Preferential Sampling)...")

for train_idx, test_idx in kf.split(X):
    # Split
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    
    # Apply Preprocessing
    X_train_pref, y_train_pref = apply_preferential_sampling(X_train, y_train)
    
    # Train
    nb_classifier.fit(X_train_pref, y_train_pref)
    
    # Predict
    y_pred = nb_classifier.predict(X_test)
    gender_test = X_test['Gender']
    
    # Metrics
    accuracies.append(accuracy_score(y_test, y_pred))
    
    # Fairness Calculations
    res = pd.DataFrame({'y_true': y_test, 'y_pred': y_pred, 'Gender': gender_test})
    
    # DP
    p1 = res[res['Gender'] == 1]['y_pred'].mean()
    p0 = res[res['Gender'] == 0]['y_pred'].mean()
    dp_diffs.append(p1 - p0)
    
    # Eq Odds
    def get_rates(df_sub):
        cm = confusion_matrix(df_sub['y_true'], df_sub['y_pred'], labels=[0, 1])
        tn, fp, fn, tp = cm.ravel()
        tpr = tp / (tp + fn) if (tp + fn) > 0 else 0
        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0
        return tpr, fpr
    
    tpr1, fpr1 = get_rates(res[res['Gender'] == 1])
    tpr0, fpr0 = get_rates(res[res['Gender'] == 0])
    
    tpr_diffs.append(tpr1 - tpr0)
    fpr_diffs.append(fpr1 - fpr0)

# ---------------------------------------------------------
# 4. Report Results
# ---------------------------------------------------------
print("-" * 40)
print(f"Average Accuracy: {np.mean(accuracies):.4f}")
print("-" * 40)
print("Group-Level Fairness Metrics (Gender 1 - Gender 0):")
print(f"Demographic Parity Difference:      {np.mean(dp_diffs):.4f}")
print(f"Equality of Odds (TPR Difference):  {np.mean(tpr_diffs):.4f}")
print(f"Equality of Odds (FPR Difference):  {np.mean(fpr_diffs):.4f}")
print("-" * 40)
```

<p style="font-size: 16px; color: blue;"><b>Authors Note Task 6:</b> Interestingly, the accuracy increased significantly compared to the baseline and previous tasks (which were around ~78.8%). This means balancing the classes (oversampling the minority cases) prevented the model from being biased towards the majority class (rejected applicants) <br> <br> <b> Demographic Parity: </b> -0.0754 - this is a big change. The negative value indicates that Gender 0 (Unprivileged) is now receiving more approvals) than Gender 1<br> <br> <b>Equality of Odds</b>: <br>
TPR Difference: ~-0.1192 (previously -0.0061)<br> This metric shows a strong swing in favor of Gender 0. A negative TPR difference of ~12% means that qualified applicants from Gender 0 are significantly more likely to be approved than qualified applicants from Gender 1. <br> <br> Preferential Sampling was extremely aggressive in this experiment.</p></p>
