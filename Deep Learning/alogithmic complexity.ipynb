```python
# Example of O(n)
arr = ['a','c','n','m']
for i in range(len(arr)):
    if arr[i] == 'm':
        print('found')
```

```python
# Example of O(n^2)

# len(matrix_2d): The list matrix_2d contains two inner lists, so its length is 2.
# len(matrix_2d[i]): This looks at the length of the list at the current index i.
      #    - When i is 0: The expression becomes len(matrix_2d[0]), which is len([1, 2, 3]). The length is 3. (0,1,2)
      #    - When i is 1: The expression becomes len(matrix_2d[1]), which is len([4, 5, 6]). The length is also 3.(0,1,2)

matrix_2d = [
    [1, 2, 3],
    [4, 5, 6]
]
# The outer loop iterates through each row of the matrix
for i in range(len(matrix_2d)): #Crange will exlude teh last value
    for j in range(len(matrix_2d[i])): # The inner loop iterates through each element (column) in the current row
        print(matrix_2d[i][j],end=' ')
```

```python
# O(log(n))

#To achieve logarithmic time complexity, the loop's control variable must be modified by a multiplicative or divisive factor, such as i *= 2 or i //= 2

# Using i += 2 in a loop does not result in \(O(\log n)\) time complexity because the number of iterations is still directly proportional to the input size, \(n\). This is the definition of linear time complexity, \(O(n)\)

# In a linear algorithm, if the input size (\(n\)) doubles, the number of operations roughly doubles as well. A loop that adds a constant amount to its counter (like i += 2) traverses the input linearly.

array_5 = [1,2,3,4,5]
i = 0
while i < len(array_5):
    print(array_5[i])
    i += 2
```
